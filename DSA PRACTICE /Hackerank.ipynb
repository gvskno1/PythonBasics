```python
n = int(input())

first = 0
second = 1
index = 1    

odd_sum = 0
odd_count = 0

while first <= n:
    print(first, end=", ")

    if index % 2 != 0:
        odd_sum += first
        odd_count += 1

    next_number = first + second
    first = second
    second = next_number
    index += 1

print()
print("sum:", odd_sum)
print("Odd rows are:", odd_count)
```
