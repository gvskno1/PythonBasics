# **1. Missing Number.**

```python
def missingNumber(self, nums: List[int]) -> int:
        n = len(nums)
        expected_sum = n*(n+1)//2
        actualSum = nums[0]
        for i in range(1,n):
            actualSum+=nums[i]
        
        return expected_sum - actualSum
```

# **2.SortedSquareArray**

```python
# using absolute 
def sortedSquares(self, nums: List[int]) -> List[int]:
        n = len(nums)
        result = [0] *n 
        left = 0
        right = n-1
        pos = n-1 
        while left <= right:
            if abs(nums[left]) > abs(nums[right]):
                result[pos] = nums[left]**2
                left+=1
            else:
                result[pos] = nums[right]**2
                right-=1
            pos -=1
        return result
```

# **3.Contains Duplicate.**

```python
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        return len(nums) != len(set(nums))
```

```python
def containsDuplicate(nums):
    seen = set()

    for n in nums:
        if n in seen:
            return True
        seen.add(n)

    return False
```

# **Container with most water**

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        n =len(height)
        left = 0
        right = n-1 
        max_water = 0

        while left < right:
            min_pillar = min(height[left], height[right])
            max_water =  max(max_water, min_pillar*(right-left))
            if height[left] < height[right]:
                left += 1
            else: 
                right -=1 
        
        return max_water



```

# **Required Count Notes**

```python
x = int(input())

notes = [2000, 500, 200, 100, 50, 20, 10, 5, 2, 1]

notes_count = {}

for note in notes:
    count = x // note        
    notes_count[note] = count
    x = x % note            

for note in notes:
    print(f"{note}: {notes_count[note]}")
```

# **Sliding Window(Variable Size)**

```python
# Given an array of positive integers. 
def minSubArrayLen(self, target: int, nums: list[int]) -> int:
    window = 0
    size = 0
    res = float('inf')
    left = 0
    n = len(nums)

    for right in range(n):
        window += nums[right]
        size += 1
        while window >= target:
            res = min(size,res)
            window -= nums[left]
            size -= 1
            left += 1
    return res if res != float('inf') else 0 



```
