```python
A,B  = input().split()
A,B = int(A), int(B)
def HCFLCM(A,B):
    a,b = A, B 

    while(B!=0):
        A,B = B,A%B
    HCF = A
    
    LCM = (a*b)//HCF
    return HCF, LCM 
HCF, LCM = HCFLCM(A,B)
print(HCF)
print(LCM)
```

```python
# Prime Signal Validator 
N = int(input())
id = [int(x) for x in input().split()]
total = id[0]
for i in range(1,N-1):
    total += id[i]
sum_of_natural= N*(N+1)//2
missing_number = sum_of_natural - total
print(missing_number)
```

```python
# Anagram Word Checker 
s1 = input()
s2 = input()
d1 = {}
d2 = {}
for ch in s1:
    if ch in d1: 
        d1[ch] +=1
    else:
        d1[ch] =1 
for ch in s2: 
    if ch in d2: 
        d2[ch] +=1 
    else: 
        d2[ch] = 1
if (d1 ==d2 ):
    print("YES")
else:
    print("NO")
```

```python
# two sum 
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)

        for i in range(n):
            for j in range(i+1,n):
                if nums[i] + nums[j] = target:
                    return [i,j]
        return []
```

```python
N = int(input())
nums = [int(x) for x in input().split()]
duplicates = []

for i in range(len(nums)):
    val = abs(nums[i])
    index = val -1 

    if nums[index] < 0:
        duplicates.append(val)
    else: 
        nums[index] = -nums[index]

if duplicates:
    duplicates = sorted(set(duplicates)) 
    for x in duplicates:
        print(x)
else:
    print("No Duplicates")
```
