# **JOINS**

```python
# A LEFT JOIN keeps all of the data from your Left table and whatever matches from the Right table.

# A RIGHT JOIN keeps all of the data from your right table and whatever matches from the Left table. 

# A FULL JOIN brings in everything from both tables and matches whatever will match from the columns you specify. 

# Inner Join in pandas ---- pd.merge(df1, df2, on='id', how = 'inner')

#  Left Join in pandas ---- pd.merge(df1, df2 , on ='id', how = 'left')

# right join in pandas ---- pd.merge(df1, df2, on ='id', how = 'right')

# Full Outer join in pandas --- pd.merge(df1, df2, on = 'id', how = 'outer')

# Join on different column names -- pd.merge(df1, df2, left_on ='user_id, right_on ='id', how ='inner')

# Union all -- pd.concat([df1, df2])

# Union -- pd.concat([df1, df2]).drop_duplicates()
```
